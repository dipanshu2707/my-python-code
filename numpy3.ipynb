```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data structure: [restaurant_id,2021,2022,2023,2024]
sales_data = np.array([
    [1,150000,180000,220000,250000], #paradise biryani
    [2,120000,140000,160000,190000], #Beijing bites
    [3,200000,210000,260000,500000], #pizza hub
    [4,180000,210000,350000,270000] #Burger point
])

# print('=== Zomato sales analysis ===')
# print('\n sales data shape', sales_data.shape)
# print('\n sample data for 1 st 3 restau:',sales_data[2])
# print('\n sample data for 1 st 3 restau:',sales_data[0:3])

print('\n sample data for 1 st 3 restau:',sales_data[:,1:])
```

```python
# Total sales per year

print(np.sum(sales_data, axis = 0))
yeary_total = np.sum(sales_data[:,1:],axis=0)
print(yeary_total)
```

```python
# Minimum sales per restaurant

min_salea = np.min(sales_data[:,1:],axis=1)
print(min_salea)
```

```python
#Maximum sales per year

mix_salea = np.max(sales_data[:,1:],axis=0)
print(mix_salea)
```

```python
# Average sales per restaurant

avg_sales = np.mean(sales_data[:,1:],axis =1)
print(avg_sales)
```

```python
cumsum = np.cumsum(sales_data[:,1:],axis = 1)
print(cumsum)
```

```python
plt.figure(figsize=(10,6))
plt.plot(np.mean(cumsum, axis=0))
plt.title("Average cumulative sales accross all restaurant")
plt.xlabel("years")
plt.ylabel("sales")
plt.grid(True)
plt.show()

```
