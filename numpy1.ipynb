### Array Reshaping

```python
arr = np.arange(12)
print('original array', arr)

```

```python
reshaped = arr.reshape((3,4))
print(" Reshape array:\n", reshaped)
```

```python
flattened = reshaped.flatten()
print("flattened array", flattened)
```

```python
 # Ravel(return view,instead of copy)
raveled = reshaped.ravel()
print("raveled array",raveled)
```

```python
#Transpose
transpose = reshaped
print("Transpose array", transpose)
```

```python
arr1 = np.arange(16)
print(arr1)
```

```python
reshaped = arr1.reshape((4,4))
print(" Reshape array:\n", reshaped)


flattened = reshaped.flatten()
print("flattened array", flattened)

raveled = reshaped.ravel()
print("raveled array",raveled)

transpose = reshaped
print("Transpose array\n", transpose)
```
