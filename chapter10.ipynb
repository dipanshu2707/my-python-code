### OBJECT ORIENTED PROGRAMMING

##### Class
A class is a blueprint or template used to create objects.

#### Object
An object is a real instance of a class.

```python
class Student:
    name = "Dipanshu"

s1 = Student()
print(s1.name)
```

```python
class Person:
    name = "Harry"
    occupation = "software developer"
    networth = 1000000

    def info(self):
        print(f"{self.name} is a {self.occupation}")

a = Person()
a.name = "Shubham"
a.occupation = "data scientist"
a.info()
```

```python
class Employee:
    name = "Ram"
    language = "Python"
    salary = 1300000

harry = Employee()
print(harry.name, harry.language)
```

```python
class Employee:
    language = "Python"   # Class attribute
    salary = 1300000

harry = Employee()
harry.name = "Harry"      # Instance attribute
print(harry.name, harry.salary, harry.language)

rohan = Employee()
rohan.name = "Rohan Roro"
print(rohan.name, rohan.salary, rohan.language)
```

```python
# Instance attributes override class attributes
class Employee:
    language = "Python"
    salary = 1300000

harry = Employee()
harry.language = "JavaScript"
print(harry.salary, harry.language)
```

```python
class Employee:
    language = "Python"
    salary = 1300000

    def getInfo(self):
        print(f"the language is {self.language}. the salary is {self.salary}")

harry = Employee()
harry.language = "JavaScript"
harry.getInfo()




```

```python
### Constructor
# A constructor is a special function that gets automatically called when an object is created.

class Person:
    def __init__(self, n, o):
        print("Hey I am a person")
        self.name = n
        self.occ = o

    def info(self):
        print(f"{self.name} is a {self.occ}")

a = Person("Harry", "software developer")
b = Person("Rohan", "HR")

a.info()
b.info()
```

```python
class Employee:
    def __init__(self, name, salary, language):
        self.name = name
        self.salary = salary
        self.language = language
        print("I am creating an object")

    def getInfo(self):
        print(f"the language is {self.language}. the salary is {self.salary}")

    @staticmethod
    def greet():
        print("Good morning")

harry = Employee("Harry", 100000, "Python")
print(harry.name, harry.salary, harry.language)
```

```python
class Programmer:
    company = "Microsoft"
    def __init__(self, name, pin, salary):
        self.name = name
        self.pin = pin
        self.salary = salary

p = Programmer("Harry", 1231, 1400000)
print(p.name, p.pin, p.salary, p.company)
r = Programmer("Rohan", 2341, 1200000)
print(r.name, r.pin, r.salary, r.company)
```

```python
class Calculator:
    def __init__(self, n):
        self.n = n

    def square(self):
        print(f"the square is {self.n*self.n}")

    def cube(self):
        print(f"the cube is {self.n*self.n*self.n}")

    def square_root(self):
        print(f"the square root is {self.n**0.5}")

    @staticmethod
    def hello():
        print("hello world")

a = Calculator(3)
a.hello()
a.square()
a.cube()
a.square_root()
```

```python
class Demo:
    a = 4

o = Demo()
print(o.a)   # class attribute
o.a = 0      # instance attribute
print(o.a)   # instance attribute
print(Demo.a) # class attribute
```

```python
from random import randint

class Train:
    def book(self, train_no, fro, to):
        print(f"Ticket is booked in train no:{train_no} from {fro} to {to}")

    def getStatus(self, train_no):
        print(f"Train no {train_no} is running on time")

    def getFare(self, train_no, fro, to):
        print(f"Fare for train no:{train_no} from {fro} to {to} is {randint(222,5555)}")

t = Train()
t.book(12345, "Delhi", "Mumbai")
t.getStatus(12345)
t.getFare(12345, "Delhi", "Mumbai")
```
